```python
import pandas as pd
import sweetviz as sv

# Cargar datos
df = pd.read_csv('heart_disease_uci.csv')

# Eliminar la columna dataset para evitar sesgo
df = df.drop(columns=['dataset'])
df = df.drop(columns=['id'])

# Crear target binario (0 = sano, 1 = enfermo)
df['target'] = (df['num'] > 0).astype(int)

# Generar reporte con target
report = sv.analyze(df, target_feat='target')

# Guardar y abrir
report.show_html("reporte.html", open_browser=True)

print("✅ Reporte generado sin la columna 'dataset'")
```
